# Synchronized double-HG GlobalFit event comparison

Compare the same MC event across every new double-HG lightmap. One event selector and one charge threshold update all selected maps simultaneously. The three panels in each row are the physical XY, XZ and YZ fibre views; magenta segments are MC truth.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
from IPython.display import display

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / 'double_hg_event_comparison_tools.py').exists():
    NOTEBOOK_DIR = Path('/home/tlux/HK/ND280++/LFGD_Recon_Simu/Studies/GlobalFitTest')
sys.path.insert(0, str(NOTEBOOK_DIR))
from double_hg_event_comparison_tools import *
plt.rcParams.update({'figure.dpi': 110})
```

## Configuration

Switch `ENERGY` between `0p7GeV` and `5GeV`, then rerun from this cell downward. Files are discovered from their `double_hg` branch names, so later scattering lengths are included automatically.

```python
ENERGY = '0p7GeV'  # '0p7GeV' or '5GeV'
BASE = Path('/home/tlux/HK/ND280++/LFGD_Recon_Simu/Studies/LightmapsStudie') / ENERGY / 'maps'
runs, truth = load_runs(BASE)
events = common_events(runs)
print(f'{ENERGY}: {len(runs)} lightmaps, {len(events)} common events, range {events[0]}..{events[-1]}')
display(pd.DataFrame([{'lightmap': label, 'scatter_length_mm': run['scatter'], 'events': run['fit'].event.nunique(), 'file': str(run['path'])} for label, run in runs.items()]))
```

## Synchronized event display

Select fewer maps for a compact display. `Shared scale` uses one charge normalization across all selected maps; disable it when one map compresses the visible range of another.

```python
interactive_comparison(runs, truth);
```

### Widget-free display

Use this if Jupyter reports `model not found`. Change `EVENT`, `THRESHOLD`, or `LABELS` and rerun the cell.

```python
EVENT = events[0]
THRESHOLD = 3.0
LABELS = list(runs)  # e.g. ['0.25 mm', '0.7 mm', '1.5 mm']
draw_synchronized_event(runs, truth, EVENT, LABELS, THRESHOLD, log_charge=True, show_mc=True, shared_scale=True);
```

## Detector-response controls

Event distributions of fibre multiplicity, summed charge, and peak-fibre charge. The returned table can be filtered event by event.

```python
event_controls, input_figure = plot_input_controls(runs)
display(event_controls.groupby('lightmap')[['hit_count', 'total_charge', 'maximum_charge']].agg(['mean', 'std', 'median']))
```

## GlobalFit controls

Status fractions, EDM, seed-to-fit angular change, and chi-square per degree of freedom. Status and column convergence are distinct; these files contain the baseline GlobalFit stage.

```python
fit_controls, fit_figure = plot_fit_controls(runs)
display(fit_controls.groupby('lightmap').agg(events=('event', 'nunique'), status0_fraction=('status', lambda x: (x == 0).mean()), median_edm=('edm', 'median'), median_chi2_ndof=('chi2_ndof', 'median'), median_seed_fit_angle_deg=('seed_fit_angle_deg', 'median')))
```
